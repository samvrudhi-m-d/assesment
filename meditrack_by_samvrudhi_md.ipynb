```python
print(df.drop_duplicates())
```

```python
df["PatientName"] = df["PatientName"].fillna("Not Provided")

df["Phone"] = df["Phone"].fillna("Not Provided")

df["StockQuantity"] = df["StockQuantity"].fillna(df["StockQuantity"].median())
```

```python
df.isnull().sum()
```

```python
df["DOB"] = df["DOB"].fillna(df["DOB"].mode()[0])

df["ExpiryDate"] = df["ExpiryDate"].fillna(df["ExpiryDate"].mode()[0])
```

```python
df.isnull().sum()
```

```python
print(df["DOB"].head(10))
print(df["DateIssued"].head(10))
print(df["ExpiryDate"].head(10))
```

```python

date_columns = ["DOB", "DateIssued", "ExpiryDate"]

for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        format="%d-%m-%Y",
        errors="coerce"
    )

print(df[date_columns].dtypes)
print(df[date_columns].head(10))
```

```python

df.to_csv("cleaned_data.csv", index=False)

from google.colab import files
files.download("cleaned_data.csv")
```
