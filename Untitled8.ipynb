```python
import numpy as np


matrix = np.random.randint(1, 50, size=(5, 5))

print("Original Matrix:")
print(matrix)



print("\nNumbers greater than 25:")
print(matrix[matrix > 25])


modified_matrix = matrix.copy()
modified_matrix[modified_matrix % 2 == 0] = 0

print("\nMatrix after replacing even numbers with 0:")
print(modified_matrix)

print("\nOriginal Matrix remains unchanged:")
print(matrix)


random_array = np.random.randint(1, 51, size=10)

minimum = random_array.min()
maximum = random_array.max()

normalized_array = (random_array - minimum) / (maximum - minimum)

print("\nOriginal 1D Array:")
print(random_array)

print("\nNormalized Array:")
print(normalized_array)
```

```python
import pandas as pd
import numpy as np

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma", "Frank", "Grace", "Henry"],
    "Department": ["IT", "HR", "IT", "Finance","HR", "Finance", "IT", "HR"
    ],
    "Age": [25, 30, 28, 35, 29, 40, 32, 27],
    "Salary": [50000, np.nan, 60000, 55000, np.nan, 70000, 65000, np.nan]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

missing_values = df.isnull().sum()
print("\nMissing values in each column:")
print(missing_values)

print("\nMissing values in each column:")
df["Salary"] = df["Salary"].fillna(
    df.groupby("Department")["Salary"].transform("median")
)

print("\nDataFrame after filling missing Salary values:")
print(df)

df["Annual_Bonus"] = df["Salary"] * 0.10

print("\nDataFrame with Annual_Bonus:")
print(df)

average_salary = df.groupby("Department")["Salary"].mean()

print("\nAverage Salary per Department:")
print(average_salary)
```

```python
import numpy as np

arr = np.arange(1, 21)

print("1D Array:")
print(arr)

matrix = arr.reshape(4, 5)

print("\n4 × 5 Matrix:")
print(matrix)

columns_2_and_3 = matrix[:, 1:3]

print("\n2nd and 3rd Columns:")
print(columns_2_and_3)

corners = np.array([
    matrix[0, 0],   # Top-left
    matrix[0, -1],  # Top-right
    matrix[-1, 0],  # Bottom-left
    matrix[-1, -1]  # Bottom-right
])

print("\nFour Corner Elements:")
print(corners)
```

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Smartphone", "Tablet", "Headphones", "Monitor", "Keyboard"],
    "Category": ["Electronics", "Electronics", "Electronics", "Accessories", "Electronics", "Accessories"],
    "Price": [75000, 45000, 30000, 1500, 12000, 2500],
    "Stock": [25, 60, 15, 45, 10, 30]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

filtered_df = df[(df["Price"] > 500) & (df["Stock"] < 50)]

print("\nProducts with Price > 500 and Stock < 50:")
print(filtered_df)
df["Restock_Needed"] = df["Stock"] < 20

print("\nDataFrame with Restock_Needed:")
print(df)

df_sorted = df.sort_values(by="Price", ascending=False)

print("\nFinal DataFrame sorted by Price (Descending):")
print(df_sorted)
```

```python
import numpy as np
celsius = np.random.uniform(15.0, 40.0, size=10)

print("Temperatures in Celsius:")
print(celsius)

fahrenheit = (celsius * 9 / 5) + 32

print("\nTemperatures in Fahrenheit:")
print(fahrenheit)


numbers = np.arange(1, 11)

square_roots = np.sqrt(numbers)

print("\nNumbers from 1 to 10:")
print(numbers)

print("\nSquare Roots:")
print(square_roots)


max_index = np.argmax(fahrenheit)

print("\nIndex of Maximum Fahrenheit Temperature:")
print(max_index)

print("\nMaximum Fahrenheit Temperature:")
print(fahrenheit[max_index])
```
